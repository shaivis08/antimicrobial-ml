```python
import pandas as pd
df = pd.read_csv('antimicrobial-ml/data/splits/y_train.csv')
df.describe()
```

```python
import matplotlib.pyplot as plt

df["Inhibition Zone (mm)"].hist(bins=20)
plt.xlabel("Inhibition Zone (mm)")
plt.ylabel("Frequency")
plt.title("Distribution of Inhibition Zone")
plt.show()
```

### Observations

- The training target contains 107 observations.
- Inhibition zone ranges from 0 to 40 mm.
- Mean inhibition zone is ~17.78 mm and median is 17 mm.
- Most observations are concentrated approximately between 10 and 25 mm.
- The distribution appears somewhat right-skewed due to a smaller number of larger values.
- The 0 mm and 40 mm observations should be checked against the original data to confirm that they represent valid measurements.
- Since inhibition zone is a continuous numerical variable, regression is currently a natural candidate formulation.
